```python
# 1. Basic data quality
print("MISSING VALUES:")
print(train_df.isnull().sum().sum())

print("\nDUPLICATE ROWS:")
print(train_df.duplicated().sum())

print("\nDUPLICATE RECORD IDs:")
print(train_df["record_id"].duplicated().sum())


# 2. Potential data leakage
print("\nPOTENTIAL LEAKAGE:")
for col in train_df.columns:
    if col not in holdout_df.columns:
        print(col, "is absent from holdout")

        
# 3. Unit mix-up
print("\nMENU PRICE BY BLOCK:")
print(train_df.groupby("block")["menu_price_rs"].agg(
    ["count", "mean", "min", "max"]
))


# 4. Sentinel values - inspect frequent extreme values
print("\nPOTENTIAL SENTINEL VALUES:")
for col in train_df.select_dtypes(include=np.number).columns:
    print(f"\n{col}:")
    print(train_df[col].value_counts().head(5))


# 5. Target typos
print("\nPOTENTIAL TARGET TYPOS:")
print(train_df[train_df["plates_sold"] >= 1000][
    ["record_id", "plates_sold"]
])


# 6. Stuck sensor - inspect consecutive repeated readings
print("\nPOTENTIAL STUCK SENSOR:")

train_sorted = train_df.copy()
train_sorted["date"] = pd.to_datetime(train_sorted["date"])
train_sorted = train_sorted.sort_values("date")

for col in [
    "students_on_campus",
    "temp_c",
    "queue_minutes_yesterday"
]:
    repeated = train_sorted[col].eq(train_sorted[col].shift())
    print(col, "Consecutive repeated readings:", repeated.sum())


# 7. Spelling variants
print("\nCATEGORICAL VALUES:")

for col in ["day_type", "block"]:
    print(f"\n{col}")
    print("TRAIN:", train_df[col].unique())
    print("LATER:", later_df[col].unique())
```

```python
# 1. Check duplicate records
duplicates = train_df[train_df.duplicated(keep=False)]

print("Duplicate records:", len(train_df[train_df.duplicated()]))
print("Unique duplicated IDs:", train_df.loc[
    train_df["record_id"].duplicated(False), "record_id"
].nunique())

# Check whether duplicate IDs have conflicting data
print("\nDuplicate ID groups with different values:")
print(
    train_df.groupby("record_id").size().loc[lambda x: x > 1].head(10)
)


# 2. Check target typos
target_suspects = train_df[train_df["plates_sold"] >= 1000]

print("\nTarget values >= 1000:", len(target_suspects))
display(target_suspects[["record_id", "plates_sold"]])


# 3. Check spelling variants
for col in ["day_type", "block"]:
    print(f"\n{col}:")
    print("Train:", sorted(train_df[col].dropna().unique()))
    print("Later:", sorted(later_df[col].dropna().unique()))


# 4. Check repeated sensor readings
print("\nRepeated sensor readings:")

sensor_cols = [
    "students_on_campus",
    "temp_c",
    "queue_minutes_yesterday"
]

for col in sensor_cols:
    print(col, train_df[col].value_counts().head(5).to_dict())


# 5. Check potential leaking columns
print("\nColumns present in train but absent in holdout:")
print(set(train_df.columns) - set(holdout_df.columns))

print("\nFeature-target correlations:")
print(
    train_df.select_dtypes(include=np.number)
    .corr()["plates_sold"]
    .sort_values(ascending=False)
)
```

```python
# 1. Spelling variants
for col in ["day_type", "block"]:
    print(f"\n{col}:")
    print("Train:", train_df[col].unique())
    print("Later:", later_df[col].unique())


# 2. Potential sentinel values
print("\nPotential sentinel values:")
for col in ["students_on_campus", "temp_c",
            "exam_days_left", "special_dishes",
            "queue_minutes_yesterday"]:
    print(col, train_df[col].value_counts().head(5).to_dict())


# 3. Stuck sensor investigation
print("\nRepeated readings over time:")

df_sorted = train_df.copy()
df_sorted["date"] = pd.to_datetime(df_sorted["date"])
df_sorted = df_sorted.sort_values("date")

for col in ["students_on_campus", "temp_c",
            "queue_minutes_yesterday"]:
    repeated = df_sorted[col].eq(df_sorted[col].shift())
    print(col, "Repeated consecutive readings:", repeated.sum())


# 4. Potential leakage candidates
print("\nPotential leakage candidates:")
print(set(train_df.columns) - set(holdout_df.columns))
```

```python
# Final checks: sentinel, stuck sensor and leakage

import pandas as pd

df = train_df.copy()
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date")

# 1. Longest consecutive repeated readings
print("LONGEST CONSECUTIVE RUNS")

for col in ["students_on_campus", "temp_c", "queue_minutes_yesterday"]:
    groups = df[col].ne(df[col].shift()).cumsum()
    runs = df.groupby(groups).agg(
        value=(col, "first"),
        count=(col, "size"),
        start_date=("date", "min"),
        end_date=("date", "max")
    )
    print(f"\n{col}:")
    print(runs.sort_values("count", ascending=False).head(3))

# 2. Extreme numeric values
print("\nEXTREME VALUES")
for col in [
    "students_on_campus", "temp_c", "exam_days_left",
    "special_dishes", "queue_minutes_yesterday"
]:
    print(f"\n{col}")
    print(df[col].value_counts().head(10))

# 3. Feature-target correlations
print("\nFEATURE-TARGET CORRELATIONS")
print(
    df.select_dtypes(include="number")
      .corr()["plates_sold"]
      .sort_values(ascending=False)
)
```
